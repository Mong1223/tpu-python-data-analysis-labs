```python
import pandas as pd

titanic_1 = pd.read_csv('drive/MyDrive/data/Titanic.csv')
titanic_2 = pd.read_csv('drive/MyDrive/data/Titanic2.csv')
titanic = pd.concat([titanic_1, titanic_2])
```

```python
print(titanic.head(10))
```

```python
print(titanic.info())
```

```python
males = titanic[titanic['Sex'].isin(['male'])]
females = titanic[titanic['Sex'].isin(['female'])]
print('males:', len(males))
print('females:', len(females))
```

```python
survived = titanic[titanic['Survived'].isin([1])]
print('survived persentage:', round(len(survived) / len(titanic), 4) * 100, '%')
```

```python
survived_males = survived[survived['Sex'].isin(['male'])]
survived_females = survived[survived['Sex'].isin(['female'])]

print('persentage survived males:', 
      round(len(survived_males) / len(survived) * 100, 4), '%')
print('persentage survived females:', 
      round(len(survived_females) / len(survived) * 100, 4), "%")
```

```python
survived_first_class = survived[survived['Pclass'] == 1]
survived_second_class = survived[survived['Pclass'] == 2]
survived_third_class = survived[survived['Pclass'] == 3 ]

print('persentage of passangers of 1st class survived:',
      round(len(survived_first_class) / len(survived) * 100, 2), '%')
print('persentage of passangers of 2nd class survived:',
      round(len(survived_second_class) / len(survived) * 100, 2), '%')
print('persentage of passangers of 3rd class survived:',
      round(len(survived_third_class) / len(survived) * 100, 2), '%')
```

```python
first_class = titanic[titanic['Pclass'] == 3]
print(round(len(first_class) / len(titanic) * 100, 2))
```

Независимые величины

```python
def get_name(full_name):
  ret = ''
  if "(" in full_name:
    ret = full_name[full_name.find('('):full_name.find(')')].split(' ')[0]
    if 'Mrs' in ret:
      ret = full_name[full_name.find('('):full_name.find(')')].split(' ')[1]
  else:
    ret = full_name[full_name.find('.'):len(full_name)].split(' ')[1]
  return ret.replace('"','')

female_names = titanic[titanic['Sex'] == 'female']['Name'].tolist()
female_names = [get_name(full_name).replace('(','').replace(')','')
                for full_name in female_names]

maybe_its_bad_idea = set(female_names)
bad_idea_too = dict()
for name in maybe_its_bad_idea:
  bad_idea_too[name] = female_names.count(name)
mostly_bad_idea = sorted(bad_idea_too.items(), key = lambda item: item[1])
bad_idea_too = {k: v for k, v in mostly_bad_idea}

print('Most popular name:', list(bad_idea_too.keys())[-1])
```
