```python
quotes = soup.find_all('div', class_ = 'col-md-6 x2')
current_page = 1
list_poetry = []
while current_page < 2:
  for quote in quotes:
    list_info = []
    author = quote.find('a', class_='black')
    name = quote.find('a', class_ ='lnkgrey')
    poetry_url = base_url + name.get('href')
    poetry_response = requests.get(poetry_url)
    poetry_soup = BeautifulSoup(poetry_response.text, 'lxml')
    poetry_text = poetry_soup.find('div', class_='col-md-9 x2').text.replace('/r','')
    list_info.append(author.text)
    list_info.append(name.text)
    list_info.append(poetry_text)
    list_poetry.append(list_info)
  current_page += 1
  response = requests.get(cathegory_url + 'p' + str(current_page) + '/')
  soup = BeautifulSoup(response.text, 'lxml')
  quotes = soup.find_all('div', class_ = 'col-md-6 x2')
```

```python
print(len(list_poetry))
with open('poetry.txt','w') as file:
  for poetry in list_poetry:
    line = poetry[0] + '-' + poetry[1] + poetry[2]
    file.write(line)
```
